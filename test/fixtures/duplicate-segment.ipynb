## 1. Kampong Ayer, Brunei

Often overlooked by travelers, Brunei's water village, Kampong Ayer, is the world's largest floating settlement. This network of wooden houses on stilts has been inhabited for over 1,300 years. Here, traditional life continues much as it has for centuries, with water taxis serving as the primary mode of transportation between the wooden walkways connecting various neighborhoods.

Local artisans still practice traditional crafts, and visitors can observe basket weaving, silver smithing, and boat building. The village even has its own schools, mosques, and fire stations, all built above the water.
